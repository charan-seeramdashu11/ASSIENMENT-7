### What are React Hooks?

In simple terms, **React Hooks** are special functions that let you "hook into" React features.

Before Hooks were introduced, you had to write complex Class components to use features like state (memory) or lifecycle methods (running code when a component appears or disappears). With Hooks, you can use these features inside simple **Functional components**.

The most common Hook is `useState`, which allows your component to remember things (like whether a button is clicked or what text is typed).

```python
%%html
<!-- We can demonstrate this directly in Colab using an HTML/React template -->
<div id="react-root"></div>

<script src="https://unpkg.com/react@18/umd/react.development.js"></script>
<script src="https://unpkg.com/react-dom@18/umd/react-dom.development.js"></script>
<script src="https://unpkg.com/@babel/standalone/babel.min.js"></script>

<script type="text/babel">
  const { useState } = React;

  function ToggleButton() {
    // Declare a state variable named 'isOn' and a function to update it 'setIsOn'
    const [isOn, setIsOn] = useState(false);

    return (
      <div style={{ fontFamily: 'Arial, sans-serif', margin: '20px' }}>
        <button
          onClick={() => setIsOn(!isOn)}
          style={{
            padding: '10px 20px',
            fontSize: '16px',
            cursor: 'pointer',
            backgroundColor: isOn ? '#4CAF50' : '#f44336',
            color: 'white',
            border: 'none',
            borderRadius: '4px'
          }}
        >
          {isOn ? 'On' : 'Off'}
        </button>
      </div>
    );
  }

  const root = ReactDOM.createRoot(document.getElementById('react-root'));
  root.render(<ToggleButton />);
</script>
```

 3 : Write a component that fetches users from
https://jsonplaceholder.typicode.com/users only once on mount and
shows how many users were loaded.

```python
%%html
<div id="user-fetch-root"></div>

<script type="text/babel">
  const { useState, useEffect } = React;

  function UserLoader() {
    const [usersCount, setUsersCount] = useState(null);
    const [loading, setLoading] = useState(true);
    const [error, setError] = useState(null);

    useEffect(() => {
      // Fetching users only once on mount (empty dependency array [])
      fetch('https://jsonplaceholder.typicode.com/users')
        .then(response => {
          if (!response.ok) {
            throw new Error('Network response was not ok');
          }
          return response.json();
        })
        .then(data => {
          setUsersCount(data.length);
          setLoading(false);
        })
        .catch(err => {
          setError(err.message);
          setLoading(false);
        });
    }, []);

    return (
      <div style={{ fontFamily: 'Arial, sans-serif', margin: '20px', padding: '15px', border: '1px solid #ddd', borderRadius: '8px', maxWidth: '300px' }}>
        <h3 style={{ marginTop: 0 }}>User Loader</h3>
        {loading && <p style={{ color: '#666' }}>Fetching users...</p>}
        {error && <p style={{ color: 'red' }}>Error: {error}</p>}
        {!loading && !error && (
          <p style={{ fontSize: '16px' }}>
            Successfully loaded: <strong>{usersCount}</strong> users.
          </p>
        )}
      </div>
    );
  }

  const root = ReactDOM.createRoot(document.getElementById('user-fetch-root'));
  root.render(<UserLoader />);
</script>
```

4 : Create a component that starts a timer with setInterval to increase a
“seconds” counter every second. Stop the timer when the component unmounts.

```python
%%html
<div id="timer-root"></div>

<script type="text/babel">
  const { useState, useEffect } = React;

  function Timer() {
    const [seconds, setSeconds] = useState(0);

    useEffect(() => {
      // Start the interval timer
      const intervalId = setInterval(() => {
        setSeconds(prevSeconds => prevSeconds + 1);
      }, 1000);

      // Return a cleanup function to clear the interval when the component unmounts
      return () => {
        clearInterval(intervalId);
        console.log("Timer stopped (unmounted)");
      };
    }, []); // Empty dependency array ensures this effect runs only once on mount

    return (
      <div style={{ fontFamily: 'Arial, sans-serif', margin: '20px', padding: '15px', border: '1px solid #ddd', borderRadius: '8px', maxWidth: '300px' }}>
        <h3 style={{ marginTop: 0 }}>Active Timer</h3>
        <p style={{ fontSize: '18px' }}>
          Seconds Elapsed: <strong>{seconds}s</strong>
        </p>
      </div>
    );
  }

  const root = ReactDOM.createRoot(document.getElementById('timer-root'));
  root.render(<Timer />);
</script>
```

6 : You have a theme value (“light” or “dark”) that is needed by a deep child
button. Show how to pass it without sending theme through every parent prop.

```python
%%html
<div id="context-root"></div>

<script type="text/babel">
  const { useState, createContext, useContext } = React;

  // 1. Create a Context with a default value
  const ThemeContext = createContext('light');

  // A deeply nested child component that needs the theme
  function DeepChildButton() {
    // 3. Consume the theme using the useContext Hook
    const theme = useContext(ThemeContext);

    const style = {
      padding: '10px 20px',
      fontSize: '14px',
      cursor: 'pointer',
      border: 'none',
      borderRadius: '4px',
      backgroundColor: theme === 'dark' ? '#333' : '#eee',
      color: theme === 'dark' ? '#fff' : '#000'
    };

    return (
      <button style={style}>
        Styled as {theme} theme
      </button>
    );
  }

  // An intermediate parent component that doesn't consume or pass down the theme prop
  function IntermediateParent() {
    return (
      <div style={{
        padding: '15px',
        border: '1px dashed #ccc',
        borderRadius: '6px',
        marginTop: '10px'
      }}>
        <p style={{ margin: '0 0 10px 0', fontSize: '14px', color: '#666' }}>
          Intermediate Component (No props passed through me!)
        </p>
        <DeepChildButton />
      </div>
    );
  }

  // Top-level component holding the state
  function ThemeApp() {
    const [theme, setTheme] = useState('light');

    const toggleTheme = () => {
      setTheme(prev => (prev === 'light' ? 'dark' : 'light'));
    };

    return (
      // 2. Wrap components in the Provider and pass the current value
      <ThemeContext.Provider value={theme}>
        <div style={{
          fontFamily: 'Arial, sans-serif',
          padding: '20px',
          border: '1px solid #ddd',
          borderRadius: '8px',
          maxWidth: '400px',
          backgroundColor: theme === 'dark' ? '#1e1e1e' : '#ffffff',
          color: theme === 'dark' ? '#ffffff' : '#000000',
          transition: 'all 0.3s ease'
        }}>
          <h3 style={{ marginTop: 0 }}>Theme Context Demo</h3>
          <button
            onClick={toggleTheme}
            style={{
              padding: '6px 12px',
              marginBottom: '10px',
              cursor: 'pointer'
            }}
          >
            Toggle Theme
          </button>
          <IntermediateParent />
        </div>
      </ThemeContext.Provider>
    );
  }

  const root = ReactDOM.createRoot(document.getElementById('context-root'));
  root.render(<ThemeApp />);
</script>
```

 7 : Make a counter with two buttons: “+1” and “-1”. Use useReducer instead
of useState.

```python
%%html
<div id="reducer-counter-root"></div>

<script type="text/babel">
  const { useReducer } = React;

  // 1. Define the initial state
  const initialState = { count: 0 };

  // 2. Define the reducer function that specifies how actions update the state
  function reducer(state, action) {
    switch (action.type) {
      case 'increment':
        return { count: state.count + 1 };
      case 'decrement':
        return { count: state.count - 1 };
      default:
        throw new Error(`Unhandled action type: ${action.type}`);
    }
  }

  function ReducerCounter() {
    // 3. Initialize useReducer with the reducer function and initial state
    const [state, dispatch] = useReducer(reducer, initialState);

    const buttonStyle = {
      padding: '10px 20px',
      fontSize: '16px',
      marginRight: '10px',
      cursor: 'pointer',
      border: 'none',
      borderRadius: '4px',
      color: 'white'
    };

    return (
      <div style={{ fontFamily: 'Arial, sans-serif', margin: '20px', padding: '15px', border: '1px solid #ddd', borderRadius: '8px', maxWidth: '300px' }}>
        <h3 style={{ marginTop: 0 }}>useReducer Counter</h3>
        <p style={{ fontSize: '18px' }}>Count: <strong>{state.count}</strong></p>
        <button
          onClick={() => dispatch({ type: 'increment' })}
          style={{ ...buttonStyle, backgroundColor: '#4CAF50' }}
        >
          +1
        </button>
        <button
          onClick={() => dispatch({ type: 'decrement' })}
          style={{ ...buttonStyle, backgroundColor: '#f44336' }}
        >
          -1
        </button>
      </div>
    );
  }

  const root = ReactDOM.createRoot(document.getElementById('reducer-counter-root'));
  root.render(<ReducerCounter />);
</script>
```

8 : Given an array nums = [1,2,3,4] and a filter value min, show how to
compute a filtered list only when nums or min change. Also pass a stable click handler
to a child list item.

```python
%%html
<div id="optimization-root"></div>

<script type="text/babel">
  const { useState, useMemo, useCallback, memo } = React;

  // A memoized child component to demonstrate how a stable callback prevents re-renders
  const ListItem = memo(({ value, onClick }) => {
    console.log(`Rendering ListItem: ${value}`);
    return (
      <li
        onClick={() => onClick(value)}
        style={{
          padding: '8px',
          margin: '4px 0',
          backgroundColor: '#eaeaea',
          borderRadius: '4px',
          cursor: 'pointer',
          listStyle: 'none'
        }}
      >
        Number: {value} (Click me)
      </li>
    );
  });

  function MemoDemoApp() {
    const [nums, setNums] = useState([1, 2, 3, 4]);
    const [min, setMin] = useState(2);
    const [otherState, setOtherState] = useState(0);

    // 1. useMemo: Only filters the list when 'nums' or 'min' changes
    const filteredNums = useMemo(() => {
      console.log('Computing filtered list...');
      return nums.filter(num => num >= min);
    }, [nums, min]);

    // 2. useCallback: Returns a stable function reference that won't change on re-renders
    const handleItemClick = useCallback((value) => {
      alert(`Clicked on item: ${value}`);
    }, []); // Empty dependency array means this function reference is created once and is completely stable

    return (
      <div style={{ fontFamily: 'Arial, sans-serif', margin: '20px', padding: '15px', border: '1px solid #ddd', borderRadius: '8px', maxWidth: '400px' }}>
        <h3 style={{ marginTop: 0 }}>useMemo & useCallback Demo</h3>

        <div style={{ marginBottom: '15px' }}>
          <label style={{ marginRight: '10px' }}>Filter (Minimum Value): </label>
          <input
            type="number"
            value={min}
            onChange={(e) => setMin(Number(e.target.value))}
            style={{ width: '50px', padding: '4px' }}
          />
        </div>

        <div style={{ marginBottom: '15px' }}>
          <button
            onClick={() => setNums([...nums, nums.length + 1])}
            style={{ padding: '6px 12px', marginRight: '10px', cursor: 'pointer' }}
          >
            Add Number
          </button>
          <button
            onClick={() => setOtherState(prev => prev + 1)}
            style={{ padding: '6px 12px', cursor: 'pointer' }}
          >
            Trigger Re-render ({otherState})
          </button>
        </div>

        <p><em>(Check the console logs. Triggering a re-render will not trigger list recalculation or list item re-renders!)</em></p>

        <ul style={{ padding: 0 }}>
          {filteredNums.map(num => (
            <ListItem key={num} value={num} onClick={handleItemClick} />
          ))}
        </ul>
      </div>
    );
  }

  const root = ReactDOM.createRoot(document.getElementById('optimization-root'));
  root.render(<MemoDemoApp />);
</script>
```

 9 : Create a controlled input for email. As the user types, show “Email:
<value>”. Why do we keep the input’s value in state?

### Why do we keep the input's value in state?

In React, keeping the input's value in state turns it into a **Controlled Component**.

1. **Single Source of Truth**: The React state acts as the single source of truth for the input's data rather than relying on the browser's DOM.
2. **Immediate UI Feedback & Validation**: Having the value immediately accessible in React state allows us to do things like real-time validation, dynamic styling, auto-formatting, or updating other parts of the interface (like displaying "Email: <value>" instantly) as the user types.
3. **Easier Form Submission**: When submitting forms, we can read directly from our React state rather than querying the DOM for input values.

```python
%%html
<div id="controlled-input-root"></div>

<script type="text/babel">
  const { useState } = React;

  function ControlledEmailInput() {
    const [email, setEmail] = useState('');

    return (
      <div style={{ fontFamily: 'Arial, sans-serif', margin: '20px', padding: '15px', border: '1px solid #ddd', borderRadius: '8px', maxWidth: '350px' }}>
        <h3 style={{ marginTop: 0 }}>Controlled Input Demo</h3>
        <div style={{ marginBottom: '15px' }}>
          <label htmlFor="email-input" style={{ display: 'block', marginBottom: '5px', fontWeight: 'bold' }}>Email Address:</label>
          <input
            id="email-input"
            type="email"
            value={email}
            onChange={(e) => setEmail(e.target.value)}
            placeholder="enter@yourdomain.com"
            style={{
              width: '100%',
              padding: '8px',
              boxSizing: 'border-box',
              borderRadius: '4px',
              border: '1px solid #ccc'
            }}
          />
        </div>
        <div style={{ fontSize: '15px', color: '#1a0dab', wordBreak: 'break-all' }}>
          <strong>Email:</strong> {email || <span style={{ color: '#aaa', fontStyle: 'italic' }}>no email entered yet</span>}
        </div>
      </div>
    );
  }

  const root = ReactDOM.createRoot(document.getElementById('controlled-input-root'));
  root.render(<ControlledEmailInput />);
</script>
```

10 : Make a button that focuses an input using a ref. Return adjacent
elements without extra wrapper nodes. In one sentence, say what React’s update
process does.

### React's Update Process

React's update process compares the newly generated Virtual DOM tree with the previous one (a process called reconciliation) and efficiently updates only the changed parts of the actual browser DOM.

```python
%%html
<div id="ref-focus-root"></div>

<script type="text/babel">
  const { useRef } = React;

  function FocusDemo() {
    // Create a ref to attach to the input element
    const inputRef = useRef(null);

    const handleFocus = () => {
      // Access the raw DOM node directly to invoke focus()
      if (inputRef.current) {
        inputRef.current.focus();
      }
    };

    return (
      <>
        <h3 style={{ fontFamily: 'Arial, sans-serif', marginTop: '20px' }}>
          useRef Focus & Fragment Demo
        </h3>
        <div style={{ fontFamily: 'Arial, sans-serif', display: 'flex', gap: '10px', alignItems: 'center' }}>
          <input
            ref={inputRef}
            type="text"
            placeholder="Click the button to focus me!"
            style={{
              padding: '8px',
              borderRadius: '4px',
              border: '1px solid #ccc'
            }}
          />
          <button
            onClick={handleFocus}
            style={{
              padding: '8px 16px',
              backgroundColor: '#673AB7',
              color: 'white',
              border: 'none',
              borderRadius: '4px',
              cursor: 'pointer'
            }}
          >
            Focus Input
          </button>
        </div>
      </>
    );
  }

  const root = ReactDOM.createRoot(document.getElementById('ref-focus-root'));
  root.render(<FocusDemo />);
</script>
```
